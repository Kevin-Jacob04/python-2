```python
age_list = [24, 25, 26, 27, 28]
name_list = ["Anu", "Kevin", "John", "David", "Arun"]
```

```python
print(age_list)
print(name_list)
```

```python
name_list.append("yazhini")
print("append name:",name_list)
```

```python
age_list.insert(2,30)
print("inserted age:",age_list)
```

```python
name_list.remove("yazhini")
print("After remove:", name_list)
```

```python
age_list.pop()
print("After pop:", age_list)
```

```python
age_list.extend([29, 30, 26])
print("After extend:", age_list)
```

```python
age_list.sort(reverse=True)
print("Descending order:", age_list)
```

```python
print("Maximum age:", max(age_list))
print("Minimum age:", min(age_list))
print("Sum of ages:", sum(age_list))
```

```python
print("First element:", name_list[0])
print("Last element:", name_list[-1])
print("Elements from index 2 to 4:", name_list[2:5])
print("Reverse order:", name_list[::-1])
```

```python
score = int(input("Enter your score (0-10): "))
```

```python
if score < 0 or score > 10:
    print("Invalid score! Enter a score between 0 and 10.")

elif score > 7:
    print("Performance: Above Average")
    print("Excellent!")

elif score >= 4:
    print("Performance: Average")
    print("Good effort! Keep practicing to improve.")

else:
    print("Performance: Below Average")
    print("Need to improve your performance. ")
```
